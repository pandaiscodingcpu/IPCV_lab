```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# defining a function to load image
def load(user_img,name):
    img = cv2.imread(user_img)
    img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    plt.imshow(img)
    plt.axis('off')
    plt.title(name)
    plt.show()
    return img
```

```python
img = load('images/cameraman.jpg','Cameraman')
```

```python
# get the negative of the image
L = img.max() # maximum intensity
L
# subtract the img from L
neg_img = L - img
plt.imshow(neg_img)
plt.axis('off')
plt.title('negative')
plt.show()
```

```python
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Camera chalu kar le...")
else:
    while True:
        ret,frame = cap.read()
        if not ret:
            print("Frame nahi mil rha hai...")
            break
        cv2.imshow('Live hai yeh...dekh!!!',frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break


cap.release()
cv2.destroyAllWindows()
```
