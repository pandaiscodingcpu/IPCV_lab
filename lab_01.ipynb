```python
import cv2
import matplotlib.pyplot as plt

# read the image using imread()
img = cv2.imread("images/tree.jpg")
if img is not None:
    # covert from bgr (default) to rgb
    img_to_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
    plt.imshow(img_to_rgb)
    plt.title("Image of a Tree")
    plt.axis("off")
    plt.show()
else:
    print("Error showing the image")
```

```python
plt.imshow(img)
plt.show()
```

```python
img_to_gray = cv2.cvtColor(img_to_rgb,cv2.COLOR_RGB2GRAY)
plt.imshow(img_to_gray)
plt.show()
```

## IMPORTANT POINTS
1. **For, BGR scale, we have Blue as 1, Green as 2 and Red as 3**  
2. **For RGB scale, we have Red as 1, Green as 2 and Blue as 3**

```python
# display each channel
b,g,br = cv2.split(img_to_rgb)
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(br,cmap='gray')
plt.title("Blue channel")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,2)
plt.imshow(g,cmap='gray')
plt.title("Green channel")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,2)
plt.imshow(b,cmap='gray')
plt.title("Red channel")
plt.axis("off")
plt.show()
```

```python
# convert bgr to gray
img_to_gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
plt.imshow(img_to_gray)
plt.title("BGR to GRAY")
plt.axis("off")
plt.show()

'''
Here, we see multicolor image not the gray one, beacuse we haven't set the cmap parameter as 'gray'
if we did so, we could have seen that gray image also
'''
```

```python
# convert bgr to gray
img_to_gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
plt.imshow(img_to_gray,cmap='gray')
plt.title("BGR to GRAY")
plt.axis("off")
plt.show()
```

```python
# cropping the image is actually very easy, since the img is a 2D numpy array we use array slicing
# but we need to keep one thing in mind that, for cropiing an image we need, x and y coordinates
# img[y1:y2,x1:x2]
crop_img = img[100:400,150:590]
img_to_rgb_crop = cv2.cvtColor(crop_img,cv2.COLOR_BGR2RGB)
plt.imshow(img_to_rgb_crop)
plt.title("Cropped Image of a Tree")
plt.axis("off")
plt.show()
```
